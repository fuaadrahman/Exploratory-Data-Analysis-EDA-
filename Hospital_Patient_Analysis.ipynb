```python
import numpy as np
import pandas as pd
```

```python
D=pd.read_csv('/content/drive/MyDrive/CSV files/Doctor.csv')
D
```

```python
P=pd.read_csv('/content/drive/MyDrive/CSV files/Patient.csv')
P
```

```python
#Find all patients assigned to doctor D01
a=pd.merge(D,P,on='doc_id',how='inner')
a.loc[a['doc_id']=='D01',['doc_name','patient_name']]
```

```python
#Find the oldest patient
b=P['age'].max()
P.loc[P['age']==b]
```

```python
#Find the total visit fees for all patients.
P['visit_fee'].sum()
```

```python
#Find doctors who have more than one patient.
b=pd.merge(D,P,on='doc_id',how='inner').groupby('doc_name')['patient_id'].count().reset_index(name='patient count')
b[b['patient count']>1]
```

```python
#Find patients whose age is greater than the average patient age.
c=P['age'].mean()
P[P['age']>c]
```

```python
#Find the average age of patients for each doctor.
P.groupby('doc_id')['age'].mean()
```

```python
#Find patients with a visit fee greater than ₹700.
P[P['visit_fee']>700]
```

```python
#Find the doctor with the most patients.
d=a.groupby('doc_name')['patient_name'].count()
d[d==d.max()]
```

```python
#Find the number of patients assigned to each doctor, including doctors with no patients
pd.merge(D,P,on='doc_id',how='left').groupby('doc_name')['patient_id'].count().reset_index(name='patient count')
```
